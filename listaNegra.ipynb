```python
from transformers import pipeline

clasificador = pipeline("sentiment-analysis", model="pysentimiento/robertuito-sentiment-analysis")
texto = "me gustan los videojuegos"
resultado = clasificador(texto)
print(resultado)
```

```python
# Crear lista negra
lista_negra = ["spam", "estafa", "odiar", "tonto"]

# Validar fucnionamiento
mensajes_de_prueba = [
    "Me encantan los videojuegos, son lo mejor",
    "Este servicio es una estafa total, no lo usen",
    "No me gusta para nada el clima de hoy, es horrible",
    "Por favor, no me envíes spam en este canal",
    "La comida de este restaurante estuvo aceptable y el servicio fue rápido"
]

def procesar_mensajes(mensajes, blacklist, modelo_clasificador):
    print("=== INICIANDO PROCESAMIENTO DE MENSAJES ===\n")

    for i, mensaje in enumerate(mensajes, 1):
        print(f"Mensaje {i}: \"{mensaje}\"")

        # 1. Validar Lista Negra
        contiene_palabra_prohibida = False
        for palabra in blacklist:
            if palabra.lower() in mensaje.lower():
                print(f"  -> EXCLUIDO: Contiene la palabra prohibida '{palabra}' (Lista Negra).")
                contiene_palabra_prohibida = True
                break

        if contiene_palabra_prohibida:
            print("-" * 50)
            continue

        # 2. Validar Análisis de Sentimiento
        prediccion = modelo_clasificador(mensaje)[0]
        sentimiento = prediccion['label']
        confianza = prediccion['score']

        if sentimiento == 'NEG':
            print(f"  -> EXCLUIDO: Sentimiento Negativo detectado (Clasificación: {sentimiento} con {confianza:.2f} de confianza).")
        else:
            print(f"  -> ACEPTADO: Mensaje apto (Clasificación: {sentimiento} con {confianza:.2f} de confianza).")

        print("-" * 50)

# Ejecutar la aplicación con los 2 filtros que hemos tenemos
procesar_mensajes(mensajes_de_prueba, lista_negra, clasificador)
```
